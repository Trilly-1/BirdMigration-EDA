```python
# Numeric columns considered for IQR-based treatment
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

# Geographic coordinates are valid across their full ranges, so exclude them from IQR capping
coordinate_cols = [
    "Start_Latitude", "Start_Longitude",
    "End_Latitude", "End_Longitude"
]
outlier_treatment_cols = [c for c in numeric_cols if c not in coordinate_cols]

def iqr_outlier_summary(data, columns):
    rows = []
    for col in columns:
        q1 = data[col].quantile(0.25)
        q3 = data[col].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        count = ((data[col] < lower) | (data[col] > upper)).sum()
        rows.append([col, q1, q3, iqr, lower, upper, count])
    return pd.DataFrame(
        rows,
        columns=["Feature","Q1","Q3","IQR","Lower_Bound","Upper_Bound","Outlier_Count"]
    ).sort_values("Outlier_Count", ascending=False)

outlier_summary = iqr_outlier_summary(df, outlier_treatment_cols)
outlier_summary
```

```python
selected_plot_cols = [
    "Flight_Distance_km", "Flight_Duration_hours",
    "Average_Speed_kmph", "Max_Altitude_m",
    "Temperature_C", "Flock_Size"
]

for col in selected_plot_cols:
    plt.figure(figsize=(8, 4))
    plt.boxplot(df[col].dropna(), vert=False)
    plt.title(f"Box Plot Before IQR Treatment: {col}")
    plt.xlabel(col)
    plt.show()
```

```python
def cap_iqr(data, columns):
    data = data.copy()
    limits = {}
    for col in columns:
        q1 = data[col].quantile(0.25)
        q3 = data[col].quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        limits[col] = (lower, upper)
        data[col] = data[col].clip(lower=lower, upper=upper)
    return data, limits

df_clean = df.copy()
df_clean, iqr_limits = cap_iqr(df_clean, outlier_treatment_cols)

print("IQR capping completed.")
print("Number of rows retained:", len(df_clean))
```

```python
drop_for_model = [
    "Bird_ID",
    "Origin",
    "Destination",
    "Recovery_Time_days",
    "Recovery_Location_Known"
]

df_model = df_clean.drop(columns=drop_for_model)

print("Original columns:", df_clean.shape[1])
print("Columns after feature selection:", df_model.shape[1])
print("\nRemoved columns:")
print(drop_for_model)
```

```python
X = df_model.drop(columns=["Migration_Success"])
y = df_model["Migration_Success"].map({
    "Failed": 0,
    "Successful": 1
})

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nTarget encoding:")
print(y.value_counts().sort_index())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
numeric_features = X_train.select_dtypes(include=np.number).columns.tolist()
categorical_features = X_train.select_dtypes(include="object").columns.tolist()

print("Numeric features:", len(numeric_features))
print(numeric_features)

print("\nCategorical features:", len(categorical_features))
print(categorical_features)
```

```python
numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, numeric_features),
        ("cat", categorical_pipeline, categorical_features)
    ],
    remainder="drop"
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()

X_train_ready = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_ready = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

print("Model-ready training data shape:", X_train_ready.shape)
print("Model-ready testing data shape:", X_test_ready.shape)

X_train_ready.head()
```
